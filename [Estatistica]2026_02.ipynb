#M**edidas de Posição**

```python
import pandas as pd
import numpy as np
from scipy.stats import trim_mean
```

```python
df = pd.read_excel('/content/[Estatistica-BCD]_DadosQuestionario_20262_(2).xlsx', sheet_name="DadosTratadosV2")

df.info()
```

```python
# Selecionar as variaveis que serão utilizadas
idade       = df['IDADE']             #Melhor medida é a Mediana
cidade      = df['C_RESIDE']          #Melhor medida é a Moda
LIVROS      = df['LIVROS']            #Melhor medida é a Mediana
PROGRAMACAO = df['EX_PROGRAMACAO']    #Melhor medida é a Moda
```

```python
#MEDIDAS DE POSIÇAO

## Média aritmética simples
media_simples1 = idade.mean() # com função nativa do python
media_simples2 = np.mean(idade) # com numpy (np)
# ':.2f' serve para fazer arredondação

print(f'A média simples 1 da idade é: {media_simples1:.1f}')

media_livros = np.mean(LIVROS)
print(f'A média livros: {media_livros:.1f}')

## Média ponderada
dados = [7.5, 8.2, 4.6, 6.0]
pesos = [0.20, 0.30, 0.35, 0.15]
pesos2 = [1, 3, 4, 2]

media_ponderada = np.average(dados, weights=pesos)
media_ponderada2 = np.average(dados, weights=pesos2)

print(f'A média ponderada é: {media_ponderada:.2f}')
print(f'A média ponderada2 é: {media_ponderada2:.2f}')

## Média aparada
media_aparada = trim_mean(idade, 0.1)
print(f'A média aparada: {media_aparada:.1f}')

## Mediana
mediana1 = idade.median()     # Com função nativa do Python
mediana2 = np.nanmedian(idade)   # Com função de numpy  # 'nan' para não tomar em conta os valores vazios (nan)

print(f'A mediana1: {mediana1:.1f}')
print(f'A mediana2: {mediana2:.1f}')

mediana_livros = np.median(LIVROS)
print(f'A mediana livros: {mediana_livros:.1f}')

## Moda
import statistics as st
moda1 = cidade.mode()
moda2 = st.mode(cidade)
# moda1 = cidade.mode() função do python, so imprime como tabela
print(f'Moda: {moda2}')

moda_prog = st.mode(PROGRAMACAO)
print(f'Moda Programaçãa: {moda_prog}')
```
