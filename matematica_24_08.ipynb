```python
nomes = {"Ana", "João", "Bruna"}
idades = {12, 15, 20}

print(idades)

lista = [1, 2, 2, 3, 3]
nome = set(nomes)

print(nome)
```

```python
#Incluir elemenos no conjunto

nomes.add("Helena") #add para incluir um único item
print("nomes", nomes)

idades.update({2, 30})   #Update para incluir vários itens - lembrar de colocar {}
print("idade", idades)
```

```python
#Excluir elemenos de um conjunto

#Usar discard() pára evitar erro se o elemento não estiver presente
nomes.discard("Ana")
print("nomes", nomes)

# Usar difference_update() para remover múltiplos elemenos
idades.difference_update({2, 30})
print("idade", idades)
```

```python
numeros = list(range(1, 21))

#Cria o conjuno A vazio
A = set()

# Verifica quais números são múliplos de 4
for numero in numeros:
  if numero % 4 == 0 and numero <= 16: #O 'and' é '&&' em java, msm caso do 'or' e '||'
    A.add(numero)

print("lista:", numeros)
print("conjunto A:", A)
```

```python
A = {"p", "q", "r"}
B = {"r", "s"}
C = {"p", "s", "t"}

a = (A.intersection (B)).union(C)
print("resultado a:", a)
b = (A.intersection(B)).intersection(C)
print("resultado b:", b)
c = (A.intersection(C)).union(B.intersection(C))
print("resultado c:", c)
```

```python
#Consruindo o diagrama de Venn

import matplotlib.pyplot as plt
from matplotlib_venn import venn2 # (se ver com 3 conjuntos venn3)

#definir os elemenos que constiuem o Digrama de Venn

A={2,4,6,8,10,12,14,16,18}
B={3,6,9,12,15,18}

A_apenas = A.difference(B)
B_apenas = B.difference(A)
A_B = A.intersection(B)

#criar o diagrama (figura) e ajusar o layout
plt.figure(figsize=(8,8))
plt.axis("off") #Isso aqui e para não mostrar a linha do eixo X e eixo Y

# Desenhar o diagrama
venn = venn2(subsets=(len(A_apenas), len(B_apenas), len(A_B)), set_labels=("Conjunto A", "Conjunto B"))

# Para não mostrar a quantidade de elementos, mas cada um dos elementos
venn.get_label_by_id("10").set_text('\n'.join(map(str, (A_apenas))))
    #"'\n'.join" para junnar os elementos e \n é para quebrar linha (colocar um embaixo do outro)
    #O map foi usado só para transformar os números em texo, porque o 'join' exige 'strings'.

venn.get_label_by_id("01").set_text('\n'.join(map(str, (B_apenas))))
venn.get_label_by_id('11').set_text('\n'.join(map(str, (A_B))))

#mostrar um ítulo
plt.title("Diagrama de Venn", fontsize=12)

#exibir o diagrama
plt.show()
```
